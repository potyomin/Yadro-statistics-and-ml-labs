# Задача 1

Пусть $X_1, X_2, \ldots, X_n$ — выборка из экспоненциального распределения с параметром $\lambda$. Найти оценку максимального правдоподобия параметра $\lambda$, сравнить ее с байесовской оценкой (MAP и математическое ожидание апостреорного распределения), подобрав сопряженное распределение. Сравнить полученные байесовские оценки с оценкой MLE. Найти предсказательное распределение

## MLE-оценка параметра $\lambda$

Правдоподобие:
$$
L(\lambda)=\prod_{i=1}^n \lambda e^{-\lambda X_i}
=\lambda^n e^{-\lambda\sum_{i=1}^n X_i}
=\lambda^n e^{-\lambda S}.
$$

Лог-правдоподобие:
$$
\ell(\lambda)=\ln L(\lambda)=n\ln\lambda-\lambda S.
$$

Производная:
$$
\ell'(\lambda)=\frac{n}{\lambda}-S.
$$

Приравниваем к нулю:
$$
\frac{n}{\lambda}-S=0 \Rightarrow \hat\lambda_{MLE}=\frac{n}{S}=\frac{1}{\bar X}.
$$


## Сопряжённое априорное распределение и апостериор

Для экспоненциального распределения по параметру $\lambda$ сопряжённым является **гамма-распределение** (в параметризации shape–rate):
$$
\lambda \sim \Gamma(\alpha,\beta),\quad
p(\lambda)=\frac{\beta^\alpha}{\Gamma(\alpha)}\lambda^{\alpha-1}e^{-\beta\lambda},\ \alpha>0,\beta>0.
$$

Апостериорное распределение:
$$
p(\lambda\mid X)\propto p(X\mid \lambda)\,p(\lambda)
\propto \left(\lambda^n e^{-\lambda S}\right)\left(\lambda^{\alpha-1}e^{-\beta\lambda}\right)
=\lambda^{\alpha+n-1}e^{-(\beta+S)\lambda}.
$$

Значит,
$$
\lambda\mid X \sim \Gamma(\alpha_n,\beta_n),
\qquad \alpha_n=\alpha+n,\quad \beta_n=\beta+S.
$$


## Байесовские оценки (MAP и апостериорное мат. ожидание) и сравнение с MLE

### 1) Апостериорное математическое ожидание
Для $\Gamma(\alpha_n,\beta_n)$:
$$
\mathbb{E}[\lambda\mid X]=\frac{\alpha_n}{\beta_n}
=\frac{\alpha+n}{\beta+S}.
$$

### 2) MAP-оценка (мода апостериора)
Для гамма-распределения мода (если $\alpha_n>1$):
$$
\hat\lambda_{MAP}=\frac{\alpha_n-1}{\beta_n}
=\frac{\alpha+n-1}{\beta+S}.
$$
(Если $\alpha+n\le 1$, максимум плотности достигается на границе $\lambda\to 0$.)

### 3) Сравнение с MLE
$$
\hat\lambda_{MLE}=\frac{n}{S}.
$$

Байесовские оценки “сглаживают” MLE за счёт априора.
Например, апостериорное среднее можно записать как взвешивание **prior mean** и **MLE**:
$$
\mathbb{E}[\lambda\mid X]=\frac{\alpha+n}{\beta+S}
=\frac{\beta}{\beta+S}\cdot\frac{\alpha}{\beta}\;+\;\frac{S}{\beta+S}\cdot\frac{n}{S}.
$$

При больших $n$ влияние априора уменьшается, и байесовские оценки приближаются к MLE.


## Предсказательное распределение и сравнение с MLE

Ищем предсказательное распределение для нового наблюдения $X_{new}$:
$$
p(x_{new}\mid X)=\int_0^\infty p(x_{new}\mid \lambda)\,p(\lambda\mid X)\,d\lambda.
$$

Здесь
$$
p(x_{new}\mid \lambda)=\lambda e^{-\lambda x_{new}},\qquad
\lambda\mid X\sim \Gamma(\alpha_n,\beta_n).
$$

Тогда интеграл берётся аналитически и даёт:
$$
p(x_{new}\mid X)=\frac{\alpha_n\,\beta_n^{\alpha_n}}{(\beta_n+x_{new})^{\alpha_n+1}},\quad x_{new}\ge 0,
$$
где $\alpha_n=\alpha+n,\ \beta_n=\beta+S$.

### Сравнение с предсказанием через MLE
Если подставлять вместо $\lambda$ оценку MLE, то предсказательная плотность будет просто экспоненциальной:
$$
p_{MLE}(x_{new})=\hat\lambda_{MLE}\,e^{-\hat\lambda_{MLE}x_{new}}.
$$

Байесовское предсказательное распределение учитывает неопределённость в $\lambda$ (через интегрирование по апостериору), поэтому по виду оно отличается от “подстановочного” варианта через MLE.


# Задача 2

**Мультиномиальное распределение**

Пусть проводится серия из $n$ испытаний и в результате каждого испытания происходит ровно одно событие из набора $A_1, A_2, \dots, A_m$, причем вероятности этих событий равны соответственно $\mathsf{p}_1, \mathsf{p}_2, \dots, \mathsf{p}_m$, причем
$$\sum_{i=1}^{m}\mathsf{p}_i = 1.$$

Тогда совместное распределение величин $X_1, X_2, \dots, X_m$, где $X_k$ — число наступлений события $A_k$ в серии из $n$ испытаний, задается вероятностями

$$
\mathsf{P}\left(X_1 = n_1, \dots, X_m = n_m, \right) = \frac{n!}{n_1!\dots n_m!}\mathsf{p}_1^{n_1}\dots \mathsf{p}_m^{n_m},
$$

где $n_1, n_2, \dots, n_m$ — произвольный набор целых неотрицательных чисел, таких что

$$\sum_{i=1}^m n_i = n.$$

Произведите байесовский вывод для мультиномиального распределения: найдите апостериорное распределение, используя в качестве сопоряженного распределения к правдоподобию [распределение Дирихле](https://ru.wikipedia.org/wiki/%D0%A0%D0%B0%D1%81%D0%BF%D1%80%D0%B5%D0%B4%D0%B5%D0%BB%D0%B5%D0%BD%D0%B8%D0%B5_%D0%94%D0%B8%D1%80%D0%B8%D1%85%D0%BB%D0%B5), найдите предсказательное распределение. # Задача 2. Мультиномиальное распределение (байесовский вывод)

Пусть проводится серия из $n$ испытаний, и в каждом испытании происходит ровно одно событие из множества $A_1,\dots,A_m$ с вероятностями $p_1,\dots,p_m$, где
$$
p_i\ge 0,\qquad \sum_{i=1}^m p_i=1.
$$
Пусть $X_k$ — число наступлений события $A_k$ в серии из $n$ испытаний. Тогда вектор $X=(X_1,\dots,X_m)$ имеет мультиномиальное распределение:
$$
\mathbb{P}(X_1=n_1,\dots,X_m=n_m)=\frac{n!}{n_1!\cdots n_m!}\,p_1^{n_1}\cdots p_m^{n_m},
\qquad \sum_{i=1}^m n_i=n.
$$

## 1) Правдоподобие
При фиксированных наблюдениях $x_i=n_i$ правдоподобие по параметрам $p=(p_1,\dots,p_m)$ (с точностью до константы, не зависящей от $p$) равно:
$$
L(p\mid X)\propto \prod_{i=1}^m p_i^{n_i}.
$$

## 2) Сопряжённый приор: распределение Дирихле
Выберем сопряжённый приор на $p$ — распределение Дирихле с параметрами $\alpha_1,\dots,\alpha_m$:
$$
p=(p_1,\dots,p_m)\sim \text{Dir}(\alpha_1,\dots,\alpha_m),
$$
его плотность (на симплексе $\{p_i\ge0,\ \sum p_i=1\}$):
$$
\pi(p)=\frac{1}{B(\alpha)}\prod_{i=1}^m p_i^{\alpha_i-1},
\qquad
B(\alpha)=\frac{\prod_{i=1}^m\Gamma(\alpha_i)}{\Gamma\!\left(\sum_{i=1}^m\alpha_i\right)}.
$$

## 3) Апостериорное распределение
По формуле Байеса:
$$
\pi(p\mid X)\propto L(p\mid X)\,\pi(p)
\propto \left(\prod_{i=1}^m p_i^{n_i}\right)\left(\prod_{i=1}^m p_i^{\alpha_i-1}\right)
=\prod_{i=1}^m p_i^{(\alpha_i+n_i)-1}.
$$
Это снова распределение Дирихле, то есть
$$
p\mid X \sim \text{Dir}(\alpha_1+n_1,\dots,\alpha_m+n_m).
$$

## 4) Байесовские оценки параметров $p_i$

### 4.1) Апостериорное математическое ожидание
Для $p\mid X\sim \text{Dir}(\alpha_1',\dots,\alpha_m')$ (где $\alpha_i'=\alpha_i+n_i$) имеем:
$$
\mathbb{E}[p_i\mid X]=\frac{\alpha_i'}{\sum_{j=1}^m\alpha_j'}
=\frac{\alpha_i+n_i}{\sum_{j=1}^m(\alpha_j+n_j)}
=\frac{\alpha_i+n_i}{\left(\sum_{j=1}^m\alpha_j\right)+n}.
$$

### 4.2) MAP-оценка (максимум апостериорной плотности)
MAP для Дирихле (при условии $\alpha_i'>1$ для всех $i$) имеет вид:
$$
\hat p_i^{MAP}=\frac{\alpha_i'-1}{\sum_{j=1}^m\alpha_j'-m}
=\frac{\alpha_i+n_i-1}{\left(\sum_{j=1}^m\alpha_j\right)+n-m}.
$$
Если для некоторого $i$ выполняется $\alpha_i'\le 1$, то максимум может достигаться на границе симплекса (некоторые компоненты могут быть равны 0).

### 4.3) Сравнение с MLE
MLE-оценка для мультиномиального распределения:
$$
\hat p_i^{MLE}=\frac{n_i}{n}.
$$
Байесовские оценки “сглаживают” частоты $n_i/n$ за счёт априора. Например, апостериорное среднее можно представить как смесь prior и данных:
$$
\mathbb{E}[p_i\mid X]=\frac{\alpha_i+n_i}{\alpha_0+n},
\qquad \alpha_0=\sum_{j=1}^m\alpha_j.
$$
При большом $n$ влияние априора уменьшается, и $\mathbb{E}[p_i\mid X]\approx \hat p_i^{MLE}$.

## 5) Предсказательное распределение (posterior predictive)
Рассмотрим следующий (новый) набор из $t$ независимых испытаний и пусть $Y=(Y_1,\dots,Y_m)$ — числа наступлений событий $A_i$ в этих $t$ испытаниях. Тогда:
$$
p(Y\mid X)=\int p(Y\mid p)\,\pi(p\mid X)\,dp.
$$
Если $Y\mid p\sim \text{Multinomial}(t,p)$ и $p\mid X\sim \text{Dir}(\alpha_1+n_1,\dots,\alpha_m+n_m)$, то предсказательное распределение — **Дирихле-мультиномиальное**:
$$
\mathbb{P}(Y_1=y_1,\dots,Y_m=y_m\mid X)=
\frac{t!}{y_1!\cdots y_m!}\,
\frac{\Gamma(\alpha_0+n)}{\Gamma(\alpha_0+n+t)}
\prod_{i=1}^m \frac{\Gamma(\alpha_i+n_i+y_i)}{\Gamma(\alpha_i+n_i)},
$$
где $\sum_{i=1}^m y_i=t$ и $\alpha_0=\sum_{i=1}^m\alpha_i$.

### Частный случай: предсказание одного испытания ($t=1$)
Если нужно предсказать вероятность, что в следующем испытании произойдёт событие $A_i$, то:
$$
\mathbb{P}(A_i\mid X)=\mathbb{E}[p_i\mid X]=\frac{\alpha_i+n_i}{\alpha_0+n}.
$$
Это и есть байесовское предсказание “следующего исхода”.
